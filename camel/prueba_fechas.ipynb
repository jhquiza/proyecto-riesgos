```python
df = pd.read_csv("Datos_2013_2025_cooperativas.csv")
df.sample(5)
```

```python
df_fecha = df[["fecha"]]
df_fecha.sample(5)
```

```python
df_fecha["fecha"] = pd.to_datetime(df_fecha["fecha"], format="%Y-%m-%d")
df_fecha.info()
```
