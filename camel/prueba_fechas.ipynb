```python
def get_first_non_unnamed_column(tmp):
    for col in tmp.columns:
        if not col.startswith("Unnamed"):
            return col
    return None

def extract_month_year(file_path):
    tmp = pd.read_excel(
    file_path,
    skiprows=1,
    nrows=0,
    )

    col = get_first_non_unnamed_column(tmp) 

    month = col.split(" ")[-2]
    if month == "DE":
        month = col.split(" ")[-3]
    year = col.split(" ")[-1]

    return month, year
```

```python
file_name = "f_puc_financiera_30112014_02022015_0_1.xls"
file_path = data_dir / file_name

print(extract_month_year(file_path))
```
